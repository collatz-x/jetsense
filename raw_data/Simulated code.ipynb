```python
import pandas as pd

# ================================================================
# CONFIG
# ================================================================
existing_path = "bronze_train.csv"   # your original 100-engine dataset
output_path   = "bronze_train_duplicated.csv"
n_duplicates  = 6                    # total copies (including original = 1 original + 5 new)

# ================================================================
# LOAD ORIGINAL DATA
# ================================================================
df = pd.read_csv(existing_path)
df = df.sort_values(["unit", "cycle"]).reset_index(drop=True)

orig_units = sorted(df["unit"].unique())
n_engines = len(orig_units)
print(f"✅ Loaded {n_engines} engines, {len(df)} rows")

# ================================================================
# DUPLICATE WITH CONSECUTIVE UNIT IDS
# ================================================================
all_dfs = [df]  # start with the original

for k in range(1, n_duplicates):
    copy_df = df.copy()
    copy_df["unit"] = copy_df["unit"] + k * n_engines
    all_dfs.append(copy_df)

df_full = pd.concat(all_dfs, ignore_index=True)
df_full = df_full.sort_values(["unit", "cycle"]).reset_index(drop=True)

# ================================================================
# SAVE OUTPUT
# ================================================================
df_full.to_csv(output_path, index=False)
print(f"✅ Duplicated dataset saved → {output_path}")
print(f"Total engines: {df_full['unit'].nunique()} | Total rows: {len(df_full)}")
```
